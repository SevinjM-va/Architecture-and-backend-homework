```python
import pandas as pd

df = pd.read_csv("/content/data/data.csv")

print(df.shape)
df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nTarget distribution:")
print(df['isFraud'].value_counts())

print("\nTarget percentage:")
print(df['isFraud'].value_counts(normalize=True))
```

```python
print("Categorical columns:")
print(df.select_dtypes(include=['object']).columns.tolist())

print("\nNumerical columns:")
print(df.select_dtypes(exclude=['object']).columns.tolist())
```

```python
from sklearn.model_selection import train_test_split


X = df.drop(columns=['isFraud'])
y = df['isFraud']


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

print("\nTrain target:")
print(y_train.value_counts())

print("\nTest target:")
print(y_test.value_counts())
```

```python
categorical_features = X_train.select_dtypes(
    include=['object']
).columns.tolist()

print("Categorical features:", categorical_features)
print("Number of categorical features:", len(categorical_features))
```

```python
X_train_cb = X_train.copy()
X_test_cb = X_test.copy()

for col in categorical_features:
    X_train_cb[col] = X_train_cb[col].fillna('missing').astype(str)
    X_test_cb[col] = X_test_cb[col].fillna('missing').astype(str)

print("Categorical NaN sayı (train):",
      X_train_cb[categorical_features].isna().sum().sum())

print("Categorical NaN sayı (test):",
      X_test_cb[categorical_features].isna().sum().sum())
```

```python
from catboost import CatBoostClassifier
from sklearn.metrics import roc_auc_score

catboost_model = CatBoostClassifier(
    verbose=100
)

catboost_model.fit(
    X_train_cb,
    y_train,
    cat_features=categorical_features
)

catboost_pred = catboost_model.predict_proba(X_test_cb)[:, 1]

catboost_auc = roc_auc_score(y_test, catboost_pred)

print("CatBoost ROC AUC:", catboost_auc)
```

```python
import xgboost as xgb
from sklearn.metrics import roc_auc_score

X_train_xgb = X_train.copy()
X_test_xgb = X_test.copy()

for col in categorical_features:
    X_train_xgb[col] = X_train_xgb[col].fillna('missing').astype('category')
    X_test_xgb[col] = X_test_xgb[col].fillna('missing').astype('category')

print(X_train_xgb.dtypes.head(20))
```

```python
xgb_model = xgb.XGBClassifier(
    enable_categorical=True,
    eval_metric='auc'
)

xgb_model.fit(
    X_train_xgb,
    y_train
)

xgb_pred = xgb_model.predict_proba(X_test_xgb)[:, 1]

xgb_auc = roc_auc_score(y_test, xgb_pred)

print("XGBoost ROC AUC:", xgb_auc)
```

```python
xgb_tuned = xgb.XGBClassifier(
    n_estimators=800,
    learning_rate=0.05,
    max_depth=6,
    min_child_weight=3,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=5,
    enable_categorical=True,
    eval_metric='auc',
    random_state=42
)

xgb_tuned.fit(
    X_train_xgb,
    y_train
)

xgb_tuned_pred = xgb_tuned.predict_proba(X_test_xgb)[:, 1]

xgb_tuned_auc = roc_auc_score(
    y_test,
    xgb_tuned_pred
)

print("Tuned XGBoost ROC AUC:", xgb_tuned_auc)
```

Была обучена модель CatBoost с параметрами по умолчанию. На тестовой выборке получено:

Для улучшения качества были изменены параметры модели: `learning_rate`, `n_estimators`, `max_depth`, `min_child_weight`, `subsample`, `colsample_bytree` и `reg_lambda`.

После настройки модели было получено:

Итоговые результаты:

| Модель                  | ROC AUC |

После настройки XGBoost значение ROC AUC увеличилось с 0.92587 до 0.93104. Таким образом, на данной тестовой выборке настроенная версия XGBoost показала более высокое значение ROC AUC, чем базовые варианты моделей.
